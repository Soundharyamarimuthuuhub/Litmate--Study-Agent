```python
def ask_with_memory(question):
    global chat_history

    results = retriever.invoke(question)
    context = "\n".join([r.page_content for r in results])

    history = "\n".join(chat_history)

    prompt = f"""You are a Tamil Nadu Class 12 English study assistant.
Answer using the textbook content.

Previous conversation:
{history}

Textbook content:
{context}

Current question:
{question}

Give a short and clear answer."""

    result = generator(prompt,max_new_tokens=80)
    answer = result[0]["generated_text"]

    chat_history.append("Student: " + question)
    chat_history.append("Assistant: " + answer)

    return answer

print(ask_with_memory("Who is the author of The Castle?"))
```

```python
def chatbot(question):
    results = retriever.invoke(question)
    context = "\n".join([r.page_content for r in results])

    prompt = f"""Answer the question using only the textbook content.
Give only one short answer.

Textbook:
{context}

Question: {question}

Answer:"""

    result = generator(prompt)
    answer = result[0]["generated_text"]

    if "Answer:" in answer:
        answer = answer.split("Answer:",1)[-1]

    answer = answer.split("\nTo summarize:",1)[0]
    answer = answer.split("\nExplanation:",1)[0]
    answer = answer.split("\nIn summary",1)[0]

    return answer.strip()

print(chatbot("Who is the author of The Castle?"))
```

```python
def chatbot(question):
    results = retriever.invoke(question)
    context = "\n".join([r.page_content for r in results])

    prompt = f"""Answer this question using the textbook.
Question: {question}
Textbook: {context}
Answer:"""

    result = generator(prompt)
    answer = result[0]["generated_text"]

    if "Answer:" in answer:
        answer = answer.split("Answer:")[-1]

    answer = answer.split("Textbook:")[0]
    answer = answer.split("\n")[0]

    return answer.strip()

print(chatbot("Who is the author of The Castle?"))
```

```python
def save_memory(question, answer):
    chat_history.append({
        "student": question,
        "agent": answer
    })

print("Memory function created successfully!")
```

```python
def ask_litmate(question):
    response = agent.invoke({
        "messages": [
            {"role": "user", "content": question}
        ]
    })

    answer = response["messages"][-1].content

    save_memory(question, answer)

    return answer

print("Litmate memory connected successfully!")
```

```python
def ask_litmate(question):
    memory_text = "\n".join(
        [f"Student: {m['student']}\nAgent: {m['agent']}" for m in chat_history]
    )

    prompt = f"""You are Litmate, a Tamil Nadu Class 12 English study assistant.

Previous conversation:
{memory_text}

Current student question:
{question}

Use the previous conversation to understand references such as "the poem", "the author", or "that chapter".
Use the textbook tools when needed.
Give a short, clear answer suitable for a Class 12 student.
"""

    response = agent.invoke({
        "messages": [
            {"role": "user", "content": prompt}
        ]
    })

    answer = response["messages"][-1].content
    save_memory(question, answer)
    return answer

print("Memory-aware Litmate created successfully!")
```
